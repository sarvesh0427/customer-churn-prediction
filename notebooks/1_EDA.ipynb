### Univariate EDA
- To understand the distribution of individual features and identify patterns, anomalies, and potential relationships with customer churn.

```python
numerical_features = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = df.select_dtypes(include=["object"]).columns.tolist()

categorical_features.remove("customerID")  # remove ID
categorical_features.remove("Churn")       # remove target
```

```python
numerical_features
```

```python
# numerical feature distribution
for col in numerical_features:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
# box plot
for col in numerical_features:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
### Categorical feature distribution
cat_cols = [
    "gender", "SeniorCitizen", "Partner", "Dependents",
    "PhoneService", "MultipleLines", "InternetService",
    "OnlineSecurity", "OnlineBackup", "DeviceProtection",
    "TechSupport", "StreamingTV", "StreamingMovies",
    "Contract", "PaperlessBilling", "PaymentMethod"
]
for col in cat_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(x=df[col])
    plt.title(f"Count of {col}")
    plt.xticks(rotation=45)
    plt.show()
```

### Bivariate EDA
- To understand how different features influence customer churn. By comparing feature values across churned and non-churned customers, we can identify key churn drivers.

```python
for col in numerical_features:
    plt.figure(figsize=(6,4))
    sns.boxplot(x="Churn", y=col, data=df)
    plt.title(f"{col} vs Churn")
    plt.show()
```

```python
for col in cat_cols:
    plt.figure(figsize=(6,4))
    churn_rate = pd.crosstab(df[col], df["Churn"], normalize="index") * 100
    churn_rate["Yes"].plot(kind="bar")
    plt.ylabel("Churn Rate (%)")
    plt.title(f"Churn Rate by {col}")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Contract vs churn
plt.figure(figsize=(6,4))
sns.countplot(x="Contract", hue="Churn", data=df)
plt.title("Contract Type vs Churn")
plt.show()
```
